```python
file_path = "../data/raw/company_sales.xlsx"

# Read both sheets from the Excel workbook.
sales = pd.read_excel(file_path, sheet_name="Sales Data")
people = pd.read_excel(file_path, sheet_name="People")
```

```python
# Check how many rows and columns each table has.
print("Sales shape:", sales.shape)
print("People shape:", people.shape)
```

```python
# Print the column names so we know exactly what information we have.
print("Sales columns:")
print(sales.columns.tolist())

print("\nPeople columns:")
print(people.columns.tolist())
```

```python
# Show the first 5 rows of the Sales table.
sales.head()
```

```python
# Show the first 5 rows of the People table.
people.head()
```

```python
# Check what type of data Python thinks each column contains.
sales.info()
```

```python
# Count how many missing values exist in each Sales column.
sales.isna().sum()
```

```python
# Count missing values in the People table.
people.isna().sum()
```

```python
# Show the different segments in the dataset.
print("Segments:")
print(sales["Segment"].unique())

# Show the different opportunity types.
print("\nOpportunity Types:")
print(sales["Opportunity Type"].unique())

# Show the different geographies.
print("\nGeographies:")
print(sales["Opp Owner Geo"].unique())

# Show the product lines.
print("\nProduct Lines:")
print(sales["Product Line"].unique())

# Show the product families.
print("\nProduct Families:")
print(sales["Product Family"].unique())
```

```python
# Count the total number of rows.
total_rows = len(sales)

# Count how many different opportunities exist.
unique_opportunities = sales["Opportunity ID"].nunique()

print("Total rows:", total_rows)
print("Unique opportunities:", unique_opportunities)
```

```python
# Count how many product-family rows belong to each opportunity.
rows_per_opportunity = sales.groupby("Opportunity ID").size()

# Show the first few results.
rows_per_opportunity.head()
```

```python
# Understand how many rows opportunities usually contain.
rows_per_opportunity.describe()
```

```python
# Find rows where every column is exactly duplicated.
duplicate_rows = sales.duplicated()

print("Number of exact duplicate rows:", duplicate_rows.sum())
```

```python
# Find all owner IDs that appear in Sales Data.
sales_owner_ids = set(sales["Opp Owner User ID"].dropna())

# Find all owner IDs that appear in the People table.
people_owner_ids = set(people["Opp Owner User ID"].dropna())

# Find Sales owners that don't have a matching person in the People table.
unmatched_owner_ids = sales_owner_ids - people_owner_ids

print("Sales owner IDs:", len(sales_owner_ids))
print("People owner IDs:", len(people_owner_ids))
print("Unmatched owner IDs:", len(unmatched_owner_ids))
```

```python
# Convert Close Date to a real datetime format.
sales["Close Date"] = pd.to_datetime(sales["Close Date"])

# Find the earliest and latest opportunity dates.
print("Earliest close date:", sales["Close Date"].min())
print("Latest close date:", sales["Close Date"].max())
```

```python
# Check the data type of Total Price.
print(sales["Total Price"].dtype)

# Show a few price values.  
print(sales["Total Price"].head())
```

### Data Quality & Initial Validation

The analysis began by validating the structure, completeness, and quality of the raw Sales and People datasets. The Sales Data contains 435 records across 10 columns, while the People table contains 9 records across 2 columns. There are 145 unique opportunities, confirming that individual opportunities can contain multiple product-family records. The dataset contains 52 exact duplicate records that required further investigation. The Sales Data contains 61 unique owner IDs, while only 9 are mapped in the People table, resulting in substantial missing representative-name information. No missing values were identified in the primary Sales Data fields aside from the incomplete owner-name mapping. Close Date is stored as a valid datetime covering March 2019 through January 2022, and Total Price is stored as a numeric field. The categorical variables include 3 owner geographies, 2 segments, 2 opportunity types, 2 product lines, and 6 product families. These checks established the primary data-quality considerations and confirmed that the dataset could proceed to structured cleaning and analytical modeling.
